# Week 1 — Environment Smoke Test

Verify the local Python environment runs Gymnasium **CartPole-v1** without errors
(M1 Week-1 task, also covers M3's baseline recording).

This notebook:
1. Prints the live CartPole observation / action spaces and checks them against the spec table
2. Runs **5 random-policy episodes** and records their lengths as the baseline

> Run top-to-bottom. A passing run ends with `smoke test passed`.

```python
import gymnasium as gym

print("gymnasium:", gym.__version__)
```

## 1. CartPole specification check

| Item | Spec |
|---|---|
| Observation | $s = [x, \dot{x}, \theta, \dot{\theta}]^T$ — 4 floats: cart position, cart velocity, pole angle, pole angular velocity |
| Action | $a \in \{0, 1\}$ — push cart left / right |
| Reward | $r_t = +1$ for every step taken |
| Episode end | pole falls past $\pm12^\circ$, cart leaves $\pm2.4$, or 500 steps reached |

```python
EXPECTED_OBS_DIM = 4       # s = [x, x_dot, theta, theta_dot]^T
EXPECTED_ACTION_DIM = 2    # a in {0, 1}

env = gym.make("CartPole-v1")

obs_dim = env.observation_space.shape[0]
action_dim = env.action_space.n

print("obs space:  ", env.observation_space)
print("action space:", env.action_space)

assert obs_dim == EXPECTED_OBS_DIM, f"obs dim {obs_dim} != {EXPECTED_OBS_DIM}"
assert action_dim == EXPECTED_ACTION_DIM, f"action dim {action_dim} != {EXPECTED_ACTION_DIM}"
print(f"spec check ok: {obs_dim} observations, {action_dim} actions")
```

## 2. Random-policy baseline — 5 episodes

Each episode uses `env.action_space.sample()` (uniform random actions).
These lengths are the baseline that trained agents must beat.

```python
lengths = []

for episode in range(5):
    obs, info = env.reset(seed=episode)
    done = False
    steps = 0
    while not done:
        action = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated
        steps += 1
    lengths.append(steps)
    print(f"episode {episode}: {steps} steps")

env.close()
```

```python
print(f"random baseline episode lengths: {lengths}")
print(f"mean: {sum(lengths) / len(lengths):.1f} steps")
print("smoke test passed")
```
